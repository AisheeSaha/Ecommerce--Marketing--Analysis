```python
df['Transaction_Date'] = pd.to_datetime(df['Transaction_Date'])                                   #changed it datetime dtypes
```

```python
df['Product_ID'] = df['Product_ID'].str.replace('Product_' , '' , regex=False);                               #Extract number-id
```

```python
df['Customer_ID'] = df['Customer_ID'].str.replace('Customer_' ,'' , regex=False)
```

```python
df.tail(5)
```

```python
df[['Customer_ID' , 'Product_ID']] = df[['Customer_ID', 'Product_ID']].astype(int)                         #changed dtypes as int
```

```python
df['Transaction_Year'] = df['Transaction_Date'].dt.year
```

```python
df['Transaction_Months'] = df['Transaction_Date'].dt.month_name()                                       #Add new month_name column
```

```python
df['Ad_CTR'] = round(df['Ad_CTR'], 2)
```
